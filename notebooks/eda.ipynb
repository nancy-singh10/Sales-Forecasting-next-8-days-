# Exploratory Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import sys
import os
sys.path.append(os.path.join(os.path.abspath(''), '../src'))

from data_preprocessing import load_and_preprocess_data
df = load_and_preprocess_data('../data/sales.csv')
```

## Trend Analysis

```python
texas = df[df['State'] == 'Texas']
plt.figure(figsize=(15,5))
plt.plot(texas.index, texas['Total'])
plt.title('Texas Sales Trend')
plt.show()
```

## Detect Seasonality using Rolling Averages

```python
texas['Total'].rolling(12).mean().plot(figsize=(15,5), title='Texas 12-Week Rolling Average')
```
